### Handling Missing Values in Total Charges

During the data-cleaning process, 11 missing values were identified in the `total_charges` column.

Further investigation revealed that all 11 affected customers had a tenure of zero months.

Since these customers had no recorded accumulated tenure, the missing total charges were replaced with zero using Pandas' `fillna()` function.

The original customer records were retained rather than deleted.

**Result:** All 11 missing values were successfully handled.
